```python
def GC_content(seq):
    g_cont = seq.count("G")
    c_cont = seq.count("C")

    gc = (g_cont + c_cont) / len(seq)

    return gc
```

```python
GC_content("AGGTGAGTACAGAT")
```

```python
print("The GC content is", GC_content("AGGTGAGTACAGAT"))
```

```python
def squared(start=3, end=9):
    results = []
    for i in range(start, end):
        r = i**2
        results.append(r)
    return results
```

```python
squared(start=1)
#Will automatically end at 8 due to range in original function
```
