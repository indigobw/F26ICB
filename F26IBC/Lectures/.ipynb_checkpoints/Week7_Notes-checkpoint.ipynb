```python
import pdb
def GC_content(seq):

    """ Write a description of your function
    >>> GC_content("ATATATATATATATAT")
    0.0
    >>> GC_content("GGGGGCCCCC")
    1.0
    >>> GC_content("ACACACTGTGTG")
    0.5
    """
    g_cont = seq.count("G")
    c_cont = seq.count("C")
    
    gc = (g_cont + c_cont) / len(seq)
    #The * symbol changed the math, it's supposed to be a /
    #pdb.set_trace()
    return(gc)

import doctest
doctest.testmod(verbose=True)
```

```python
seq = "AAAGCGTTTCGC"

GC_content(seq)
```

```python
import re

dna = "ACAAAATACGTTTTGAAATGTTGTGCTGTTAACACTGTCGACTAAACTTGGTAGCAAACACTTCCAAAAGGAATTCACCGGTTTCCAAAGACAGTCTTCTAATTCCTCATTAGTAATAAGTAAAATGTTTATTGTTGTAGCTCTGGACCGGTTTCCAAAGACAGTCTTCTAATTCCTCATTAGTAATAAGTAAAATCTTTATTCTTGTATACCTGG"

#r stops new lines from being made
match = re.search(r"GAATTC", dna)

match.group()
```

```python
import re

dna = "ACAAAATACGTTTTGAAATGTTGTGCTGTTAACACTGTCGACTAAACTTGGTAGCAAACACTTCCAAAAGGAATTCACCGGTTTCCAAAGACAGTCTTCTAATTCCTCATTAGTAATAAGTAAAATGTTTATTGTTGTAGCTCTGGACCGGTTTCCAAAGACAGTCTTCTAATTCCTCATTAGTAATAAGTAAAATCTTTATTCTTGTATACCTGG"

#r stops new lines from being made, re.search gives the ONLY the first variable, re.findall gives all
match = re.findall(r"TTC", dna)

len(match)
```

```python
import re

dna = "ACAAAATACGTTTTGAAATGTTGTGCTGTTAACACTGTCGACTAAACTTGGTAGCAAACACTTCCAAAAGGAATTCACCGGTTTCCAAAGACAGTCTTCTAATTCCTCATTAGTAATAAGTAAAATGTTTATTGTTGTAGCTCTGGACCGGTTTCCAAAGACAGTCTTCTAATTCCTCATTAGTAATAAGTAAAATCTTTATTCTTGTATACCTGG"

match = re.findall(r"GT[AC][GT]AC", dna)

print(match)
```

```python
mRNA = "TGCAAACTCTGAGGGCAGCAAAAAACATGAGAAAAAAAAAA"
if re.search(r"A{7,}", mRNA):
    print("It's an mRNA!")
else:
    print("No poly-A tail")
```

```python
mRNA2 = "TGGTCTATCGTAGAGTATCGGATAAAAAAAAGCGTAGTGAT"
if re.search(r"A{4,}$", mRNA2):
    print("It's an mRNA!")
else:
    print("No poly-A tail")
```

```python
xMil = re.compile(r"GT[AC][GT]AC")

seq = "TCCGATTGCTGTAGACGGTTCCGTCGACAAAAAATTT"

re.split(xMil, seq)
```

```python
hits = re.finditer(xMil, seq)

for i in hits:
    print(i.start() + 1, i.group())
    # + 1 allows for the first T to be the first nucleotide
```
