```python
x = 0
found = 0
while found < 100:
    x = x + 1
    if x % 3 == 0:
        print ("Fizz")
        continue
    elif x % 5 == 0:
        print("Buzz")
        found = found + 1
```

```python
x = 0
found = 0
while found < 30:
    x = x + 1
    if x % 3 == 0:
        print ("Fizz")
        continue
    if x % 5 == 0:
        print("Buzz")
    else:
        print(x)
    found = found + 1
```

```python
#Adding an if statement for Fizz and Buzz numbers
```

```python
x = 0
found = 0
while found < 75:
    x = x + 1
    if x % 3 == 0:
        print ("Fizz")
        continue
    elif x % 5 == 0:
        print("Buzz")
    elif x % 15 == 0:
        print("FizzBuzz")
    else:
        print(x)
    found = found + 1
```

```python
#Finding the right number of founds and reordering the if statements to get final result
```

```python
x = 0
found = 0
while found < 67:
    x = x + 1
    if x % 15 == 0:
        print("FizzBuzz")
    if x % 3 == 0:
        print ("Fizz")
        continue
    if x % 5 == 0:
        print("Buzz")
    else:
        print(x)
    found = found + 1
```
